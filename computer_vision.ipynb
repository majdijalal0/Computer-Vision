    <p style="color: #1D4ED8; font-weight: 700;">
        Question 1: What does the matrix returned by OpenCV represent?
    </p>
    <p style="color: #334155; line-height: 1.8;">
        The matrix returned by <code>cv2.imread()</code> represents the
        image as an array of pixel values. For a color image, it generally
        has three channels ordered as Blue, Green, and Red (BGR). Each pixel
        contains intensity values ranging from 0 to 255.
        The matrix dimensions correspond to the image height, width, and
        number of channels.
    </p>
    <p style="color: #1D4ED8; font-weight: 700;">
        Question 2: Why do we often convert an image to grayscale before processing?
    </p>
    <p style="color: #334155; line-height: 1.8;">
        Grayscale conversion reduces the image from three color channels
        to a single intensity channel. This simplifies computations, reduces
        the amount of data to process.
    </p>
    <p style="color: #1D4ED8; font-weight: 700;">
        Question 3: What are the benefits of applying a Gaussian blur?
    </p>
    <p style="color: #334155; line-height: 1.8;">
        A Gaussian blur smooths an image by reducing high-frequency
        variations and small details, including some image noise. It
        calculates pixel values using a Gaussian-weighted neighborhood.
        This can improve the reliability of  operations, such
        as Canny edge detection, by reducing unwanted edges caused by noise.
        However, excessive blurring can remove important details and
        weaken edges.
    </p>
</div>

```python
#We will turn the video to gray
import cv2

cap = cv2.VideoCapture('videos/parc.mp4')

while cap.isOpened():
    ret, frame = cap.read()

    if not ret:
        break

    #We apply it for every frame
    gray = cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)

    #Display the gray frame
    cv2.imshow('Gray Video', gray)

    #For stopping the video
    if cv2.waitKey(25) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
#Saving the new gray video
cap = cv2.VideoCapture('videos/parc.mp4')
w = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
h = int (cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter('video/gray_parc.mp4',fourcc,20.0,(w,h),isColor=False)

while True:
    ret , frame = cap.read()
    if not ret:
        break
    gray = cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
    out.write(gray)
    cv2.imshow('Gray Parc',gray)
    if cv2.waitKey(25) & 0xFF == ord('q'):
        break
cap.release()
out.release()
cv2.destroyAllWindows()
```

```python
import cv2
import torch
from detectron2.engine import DefaultPredictor
from detectron2.config import get_cfg
from detectron2 import model_zoo
from detectron2.utils.visualizer import Visualizer
from detectron2.data import MetadataCatalog
from google.colab.patches import cv2_imshow
# Configuration du modèle pré-entraîné
CONFIG = "COCO-InstanceSegmentation/mask_rcnn_R_50_FPN_3x.yaml"
cfg = get_cfg()
cfg.merge_from_file(model_zoo.get_config_file(CONFIG))
cfg.MODEL.ROI_HEADS.SCORE_THRESH_TEST = 0.5
cfg.MODEL.WEIGHTS = model_zoo.get_checkpoint_url(CONFIG)
cfg.MODEL.DEVICE = "cpu"
predictor = DefaultPredictor(cfg)
# Charger une image
img = cv2.imread(filename)
# Prédiction
outputs = predictor(img)
# Visualisation
v = Visualizer(img[:, :, ::-1],
MetadataCatalog.get(cfg.DATASETS.TRAIN[0]), scale=1.2) # BGR -> RGB
out = v.draw_instance_predictions(outputs["instances"].to("cpu"))
```

```python
import numpy as np
def calculate_ear(eye):
    p1, p2, p3, p4, p5, p6 = np.array(eye, dtype=float)

    vertical1 = np.linalg.norm(p2 - p6)
    vertical2 = np.linalg.norm(p3 - p5)
    horizontal = np.linalg.norm(p1 - p4)

    return (vertical1 + vertical2) / (2.0 * horizontal)

for landmarks in landmarks_list:
    left_ear = calculate_ear(landmarks["left_eye"])

print(left_ear)
```

```python
import time
import cv2
from ultralytics import YOLO
from google.colab.patches import cv2_imshow
```

```python
model_yolo = YOLO("yolov8n.pt")


model_yolo.predict(filename, conf=0.5, verbose=False)

#Starting the timer
start = time.perf_counter()
#Making predictions of objects using the model we defined , we use a confidence of 0.5
results_yolo = model_yolo.predict(
    filename, conf=0.5, verbose=False
)
#Calculating the time of execution
yolo_time = time.perf_counter() - start

print(f"YOLOv8n inference time: {yolo_time * 1000:.2f} ms")
cv2_imshow(results_yolo[0].plot())
```

```python
from detectron2.engine import DefaultPredictor
from detectron2.config import get_cfg
from detectron2 import model_zoo
from detectron2.utils.visualizer import Visualizer
from detectron2.data import MetadataCatalog

#Defining the model we are going to be using : Mask R-CNN
config_name = "COCO-InstanceSegmentation/mask_rcnn_R_50_FPN_3x.yaml"

#Defining the configuration
cfg = get_cfg()
cfg.merge_from_file(model_zoo.get_config_file(config_name))
cfg.MODEL.WEIGHTS = model_zoo.get_checkpoint_url(config_name)
cfg.MODEL.ROI_HEADS.SCORE_THRESH_TEST = 0.5 #0.5 threshold

cfg.MODEL.DEVICE = "cuda" if cv2.cuda.getCudaEnabledDeviceCount() > 0 else "cpu"

#Initializing the Mask R-CNN model
predictor = DefaultPredictor(cfg)
img = cv2.imread(filename)


predictor(img)

#Starting the timer
start = time.perf_counter()
outputs = predictor(img)
#Calculating execution time
mask_rcnn_time = time.perf_counter() - start

print(f"Mask R-CNN inference time: {mask_rcnn_time * 1000:.2f} ms")

#Displaying the image with the predictions
visualizer = Visualizer(
    img[:, :, ::-1],
    metadata=MetadataCatalog.get(cfg.DATASETS.TRAIN[0]),
    scale=1.0
)
result_image = visualizer.draw_instance_predictions(
    outputs["instances"].to("cpu")
)
cv2_imshow(result_image.get_image()[:, :, ::-1])
```

```python
#Using segmentation
model = YOLO("yolov8n-seg.pt")

try:
    results = model(filename, conf=0.5)

    #Display the image with segmentation masks
    annotated_img = results[0].plot()
    cv2_imshow(annotated_img)

    #Check whether masks were generated
    if results[0].masks is not None:
        print("Segmentation masks detected:",
              len(results[0].masks))
    else:
        print("No segmentation masks detected.")
except Exception as e:
    print("Error during segmentation:", e)
```

      Yes, YOLO can provide segmentation masks using a segmentation model such as yolov8n-seg.pt. Unlike yolov8n.pt, which detects objects using bounding boxes, the segmentation model generates a mask for each detected object.
    </p>
    <p style="color: #1D4ED8; font-weight: 700;">
        Question 3: When should we use each model?
    </p>
    <p style="color: #334155; line-height: 1.8;">
     We use YOLOv8n when we need fast object detection with bounding boxes , for example counting cars in a street or detecting pedestrians in real time.
     We use YOLOv8n-seg when we need to identify the exact visible shape of each object using segmentation masks , for example separating people from the background or outlining individual objects.
     We use Mask R-CNN when we need instance segmentation and are willing to accept potentially slower inference for detailed predictions , for example medical-image analysis, detailed object extraction, or complex scenes.

### Each person now is affected an ID, so we can keep track of the same person and verify him crossing the virutal line

```python
import cv2
from ultralytics import YOLO

cap = cv2.VideoCapture(path)

FRAME_SKIP = 1       # set to 2 for more speed (process every 2nd frame)
DEAD_ZONE = 4        # pixels around the line where we ignore jitter
IGNORE_ZONES = []    # e.g. [(0, 0, 420, 350)] to ignore the shop window (x1, y1, x2, y2)

ok, first = cap.read()

#Checking if the video is readable
if not ok:
    raise SystemExit("Cannot read the video")
    
#Getting the height and width of the video    
H, W = first.shape[:2]
print("Frame size:", W, H)

#We define our virutal line position by cliking on two points at beginning 
points = []
def on_click(event, x, y, flags, param):
    if event == cv2.EVENT_LBUTTONDOWN and len(points) < 2:
        points.append((x, y)) #We append the two points chosen 

cv2.namedWindow("People Counter", cv2.WINDOW_NORMAL)
cv2.resizeWindow("People Counter", min(W, 1280), int(min(W, 1280) * H / W))
cv2.setMouseCallback("People Counter", on_click) #Calling the function that allow us to choose points


while len(points) < 2:
    #We only display the first frame until the user choses the points of the virutal line
    preview = first.copy()
    for p in points:
        #We draw a circle for each point defining the line
        cv2.circle(preview, p, 6, (0, 255, 0), -1)
    #Add text to instruct the user to add the two points by cliking
    cv2.putText(preview, "Click the 2 ends of the line (left wall -> right edge)",
                (20, 40), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 255, 0), 2)
    cv2.imshow("People Counter", preview)
    if cv2.waitKey(30) & 0xFF == ord("q"):
        raise SystemExit
        
#A and B are the start and end of the line
A, B = points
cap.set(cv2.CAP_PROP_POS_FRAMES, 0)   #go back to the first frame

#The dimensions of the line
dx, dy = B[0] - A[0], B[1] - A[1]
#The length of the line
length = (dx * dx + dy * dy) ** 0.5

#The function helps us calculate the distance from a point to the line , we will use it to keep track of the distance between a person and the line
def signed_distance(p):
    return (dx * (p[1] - A[1]) - dy * (p[0] - A[0])) / length

def within_segment(p, margin=0.05):
    """True if p projects onto the drawn segment (not its infinite extension)."""
    t = ((p[0] - A[0]) * dx + (p[1] - A[1]) * dy) / (length * length)
    return -margin <= t <= 1 + margin

def in_ignore_zone(p):
    #Cheking if the point is in the zones ignored that we can defined as the stores on the left 
    return any(x1 <= p[0] <= x2 and y1 <= p[1] <= y2 for x1, y1, x2, y2 in IGNORE_ZONES)

#Counting the people crossing the line
count = 0
#Storing info from last frame
last_side = {}
#The ids passing the line should be unique
counted_ids = set()
frame_idx = 0

#Loop reading all frames
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break
    #we keep track of the number of the current frame
    frame_idx += 1
    #We skip a frame 
    if frame_idx % FRAME_SKIP != 0:
        continue
        
    #Using our model to keep track of people with their ids using YOLO and bytetrack
    results = model_yolo.track(
        frame,
        persist=True,
        tracker="bytetrack.yaml",
        classes=[0],
        conf=0.4,
        imgsz=640,
        verbose=False,
    )
    #The boxes that surround people
    boxes = results[0].boxes
    
    if boxes is not None and boxes.id is not None:
        ids = boxes.id.cpu().numpy().astype(int)
        coords = boxes.xyxy.cpu().numpy()
        
        #We get the coordinates of the boxes to calculate the position of the foot
        for pid, (x1, y1, x2, y2) in zip(ids, coords):
            #Position of the foot
            foot = (int((x1 + x2) / 2), int(y2))

            if in_ignore_zone(foot):
                continue

            d = signed_distance(foot)
            if abs(d) > DEAD_ZONE:
                side = 1 if d > 0 else -1
                prev = last_side.get(pid)

                if (prev is not None and prev != side and pid not in counted_ids and within_segment(foot)):
                    #increase the count if the id of the person hasn't been counted before and if he is moving and not just standing on the line 
                    count += 1
                    #Adding the person id to the counted ids to not count him again
                    counted_ids.add(pid)
                    
                #updating the position of the person
                last_side[pid] = side

            #Red circle on the foot of the person
            cv2.circle(frame, foot, 5, (0, 0, 255), -1)
            #Text displaying the id of the person next to his foot
            cv2.putText(frame, str(pid), (foot[0] + 6, foot[1]),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255, 255, 0), 1)

    #Drawing the virutal line with the points we have chosen
    cv2.line(frame, A, B, (0, 255, 0), 3)
    
    for x1, y1, x2, y2 in IGNORE_ZONES:
        cv2.rectangle(frame, (x1, y1), (x2, y2), (255, 0, 0), 1)
        
    #Displaying the count on top left with a green font
    cv2.putText(frame, f"People Count: {count}", (30, 50),
                cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)

    cv2.imshow("People Counter", frame)
    #Closing when clicking on q
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
print("Total people counted:", count)
```
